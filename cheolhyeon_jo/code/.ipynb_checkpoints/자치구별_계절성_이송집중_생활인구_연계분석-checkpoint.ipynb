# 자치구별 계절성 이송 집중도 + 생활인구 연계 분석

## 목적
1. **특정 계절에 이송건수가 상대적으로 집중되는 자치구는 어디인가?**
2. 그 집중 현상이 여러 연도에 걸쳐 반복되는가?
3. 해당 계절에 **생활인구도 함께 증가하는가?**
4. 생활인구 변화만으로 설명되지 않는 계절성 이송 증가가 있는가?

계절 정의:
- 봄: 3~5월
- 여름: 6~8월
- 가을: 9~11월
- 겨울: 12~2월

```python
# 0. 라이브러리
import os
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

plt.rc('font', family='Malgun Gothic')
plt.rc('axes', unicode_minus=False)
sns.set_theme(style='whitegrid')
pd.set_option('display.max_columns', 100)

SEASON_ORDER = ['봄', '여름', '가을', '겨울']
SEASON_MAP = {
    1:'겨울', 2:'겨울', 3:'봄', 4:'봄', 5:'봄', 6:'여름',
    7:'여름', 8:'여름', 9:'가을', 10:'가을', 11:'가을', 12:'겨울'
}
```

```python
# 1. 데이터 로드
TRANSPORT_FILE = 'C:/ai/source/09_Project/cheolhyeon_jo/ml_data/transport_dataset.csv'
POPULATION_FILE = 'C:/ai/source/09_Project/cheolhyeon_jo/ml_data/population_dataset.csv'

transport = pd.read_csv(TRANSPORT_FILE)
population = pd.read_csv(POPULATION_FILE)

transport['ds'] = pd.to_datetime(transport['ds'])
population['ds'] = pd.to_datetime(population['ds'])

print('이송 데이터:', transport.shape, transport['ds'].min(), '~', transport['ds'].max())
print('생활인구 데이터:', population.shape, population['ds'].min(), '~', population['ds'].max())
print('자치구 수:', transport['District'].nunique())
```

## 1. 자치구 × 연도별 계절 비중

연도별 데이터 누락 때문에 계절 비중이 왜곡되지 않도록 **12개월이 모두 존재하는 자치구-연도만 사용**합니다.

각 연도에서:

`계절 집중 비중 = 해당 계절 이송건수 / 해당 연도 전체 이송건수`

를 계산합니다.


```python
# 2. 완전한 12개월을 가진 자치구-연도만 선택
t = transport.dropna(subset=['transport_cnt']).copy()
t['연도'] = t['ds'].dt.year
t['월'] = t['ds'].dt.month
t['계절'] = t['월'].map(SEASON_MAP)

month_count = (
    t.groupby(['District', '연도'])['월']
     .nunique()
)
valid = month_count[month_count == 12].reset_index()[['District','연도']]

t_full = t.merge(valid, on=['District','연도'], how='inner')

print('사용 가능한 완전 자치구-연도:', len(valid))
display(valid.groupby('District').size().describe())
```

```python
# 3. 연도별 계절 비중
annual_total = (
    t_full.groupby(['District','연도'])['transport_cnt']
          .sum()
          .rename('연간이송')
)

season_total = (
    t_full.groupby(['District','연도','계절'])['transport_cnt']
          .sum()
          .rename('계절이송')
)

season_share = (season_total / annual_total).rename('계절비중').reset_index()

# 자치구별 여러 연도의 평균
district_season = (
    season_share.groupby(['District','계절'])['계절비중']
                .agg(['mean','std','count'])
                .reset_index()
                .rename(columns={
                    'mean':'평균계절비중',
                    'std':'표준편차',
                    'count':'사용연도수'
                })
)

pivot = district_season.pivot(
    index='District', columns='계절', values='평균계절비중'
)[SEASON_ORDER]

pivot['최대집중계절'] = pivot.idxmax(axis=1)
pivot['최대계절비중'] = pivot[SEASON_ORDER].max(axis=1)
pivot['25%초과폭'] = pivot['최대계절비중'] - 0.25

result = pivot.sort_values('최대계절비중', ascending=False)

print('=== 특정 계절 집중도가 높은 자치구 ===')
display((result.head(15) * 100).round(2) if False else result.head(15).round(4))
```

```python
# 보기 좋은 % 표
display_table = result.copy()
for c in SEASON_ORDER + ['최대계절비중','25%초과폭']:
    display_table[c] = (display_table[c] * 100).round(2)

display_table = display_table.rename(columns={
    '봄':'봄(%)', '여름':'여름(%)', '가을':'가을(%)', '겨울':'겨울(%)',
    '최대계절비중':'최대계절비중(%)',
    '25%초과폭':'25%기준초과(%p)'
})

display(display_table.head(15))
```

## 2. 자치구 × 계절 히트맵

25%보다 높은 계절은 그 자치구의 연간 이송 중 상대적으로 많은 비중을 차지한다는 뜻입니다.


```python
# 4. 히트맵
heat = pivot[SEASON_ORDER].copy() * 100
heat = heat.loc[result.index]

plt.figure(figsize=(9, 11))
sns.heatmap(
    heat,
    annot=True,
    fmt='.1f',
    center=25,
    cmap='RdYlBu_r',
    cbar_kws={'label':'연간 이송건수 중 계절 비중(%)'}
)
plt.title('자치구별 계절 이송 집중도')
plt.xlabel('계절')
plt.ylabel('자치구')
plt.tight_layout()
plt.show()
```

## 3. 집중도가 높은 자치구 TOP 10

각 자치구에서 가장 높은 계절 비중만 가져와 비교합니다.


```python
# 5. 최대 계절 집중도 TOP 10
top10 = result.head(10).reset_index()

plt.figure(figsize=(10,6))
sns.barplot(data=top10, x='최대계절비중', y='District', hue='최대집중계절', dodge=False)
plt.axvline(0.25, linestyle='--', label='균등 기준 25%')
plt.title('특정 계절 이송 집중도 TOP 10 자치구')
plt.xlabel('해당 계절의 연간 이송 비중')
plt.ylabel('자치구')
plt.legend(title='집중 계절')
plt.tight_layout()
plt.show()

display(top10[['District','최대집중계절','최대계절비중','25%초과폭']].round(4))
```

## 4. 반복성 확인

평균 비중이 높더라도 특정 한두 해 때문에 생긴 결과일 수 있습니다.

따라서 각 자치구의 **최대 집중 계절이 연도별로 몇 번 1위였는지** 확인합니다.


```python
# 6. 연도별 최대 계절과 반복성
year_pivot = season_share.pivot_table(
    index=['District','연도'], columns='계절', values='계절비중'
)[SEASON_ORDER]

year_winner = year_pivot.idxmax(axis=1).rename('연도별최대계절').reset_index()

repeat_rows = []
for district, row in result.iterrows():
    target = row['최대집중계절']
    tmp = year_winner[year_winner['District'] == district]
    repeat_rows.append({
        'District': district,
        '대표집중계절': target,
        '완전연도수': len(tmp),
        '해당계절1위연도수': (tmp['연도별최대계절'] == target).sum(),
        '반복률': (tmp['연도별최대계절'] == target).mean()
    })

repeat_df = pd.DataFrame(repeat_rows).sort_values(
    ['반복률','해당계절1위연도수'], ascending=False
)

display(repeat_df.head(15))
```

## 5. 생활인구 월 단위 데이터 생성

`population_dataset.csv`는 일 단위이므로 월 단위로 집계합니다.

여기서는 현재 시점에서 직접 사용할 수 있는 `일최대인구수`의 월평균을 대표 생활인구 지표로 사용합니다.


```python
# 7. 생활인구를 월 단위로 집계
p = population.copy()
p['ym'] = p['ds'].dt.to_period('M').dt.to_timestamp()

pop_monthly = (
    p.groupby(['District','ym'])['일최대인구수']
     .mean()
     .reset_index(name='월평균_일최대인구')
)

transport_monthly = transport[['District','ds','transport_cnt']].copy()
transport_monthly = transport_monthly.rename(columns={'ds':'ym'})

merged = transport_monthly.merge(
    pop_monthly, on=['District','ym'], how='inner'
).dropna()

merged['월'] = merged['ym'].dt.month
merged['계절'] = merged['월'].map(SEASON_MAP)

print('결합 데이터:', merged.shape)
print('결합 기간:', merged['ym'].min(), '~', merged['ym'].max())
display(merged.head())
```

## 6. 계절별 이송지수와 생활인구지수

자치구마다 규모가 다르므로 원자료를 바로 비교하지 않고 각 자치구의 전체기간 평균을 100으로 놓습니다.

- `이송지수 108` → 해당 계절 이송건수가 그 구의 평균보다 약 8% 높음
- `인구지수 102` → 해당 계절 생활인구가 그 구의 평균보다 약 2% 높음

두 지수를 같이 보면 **이송 증가가 생활인구 증가와 함께 나타나는지** 확인할 수 있습니다.


```python
# 8. 계절별 이송/인구 지수
season_mean = (
    merged.groupby(['District','계절'])
          .agg(
              평균이송=('transport_cnt','mean'),
              평균생활인구=('월평균_일최대인구','mean')
          )
          .reset_index()
)

overall_mean = (
    merged.groupby('District')
          .agg(
              전체평균이송=('transport_cnt','mean'),
              전체평균생활인구=('월평균_일최대인구','mean')
          )
          .reset_index()
)

season_index = season_mean.merge(overall_mean, on='District')
season_index['이송지수'] = season_index['평균이송'] / season_index['전체평균이송'] * 100
season_index['생활인구지수'] = season_index['평균생활인구'] / season_index['전체평균생활인구'] * 100
season_index['이송-인구지수차'] = season_index['이송지수'] - season_index['생활인구지수']

focus = (
    result[['최대집중계절']]
    .reset_index()
    .merge(
        season_index,
        left_on=['District','최대집중계절'],
        right_on=['District','계절'],
        how='left'
    )
    .sort_values('이송지수', ascending=False)
)

display(
    focus[['District','최대집중계절','이송지수','생활인구지수','이송-인구지수차']]
    .round(2)
    .head(15)
)
```

```python
# 9. 집중 계절의 이송 증가와 생활인구 증가 비교
plot_focus = focus.head(15).copy()

plot_long = plot_focus.melt(
    id_vars=['District','최대집중계절'],
    value_vars=['이송지수','생활인구지수'],
    var_name='지표',
    value_name='지수'
)

plt.figure(figsize=(12,6))
sns.barplot(data=plot_long, x='District', y='지수', hue='지표')
plt.axhline(100, linestyle='--')
plt.title('계절 집중 상위 자치구: 이송 증가와 생활인구 변화 비교')
plt.xlabel('자치구')
plt.ylabel('자치구 전체 평균 = 100')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 7. 생활인구와 이송건수의 전체/자치구 내부 관계

자치구 자체의 크기 때문에 상관계수가 높아지는 효과를 줄이기 위해,
각 자치구 평균을 뺀 뒤에도 생활인구와 이송건수가 함께 움직이는지 확인합니다.


```python
# 10. 전체 상관관계와 자치구 내부 상관관계
raw_r, raw_p = stats.pearsonr(
    merged['월평균_일최대인구'],
    merged['transport_cnt']
)

within = merged.copy()
within['이송편차'] = (
    within['transport_cnt']
    - within.groupby('District')['transport_cnt'].transform('mean')
)
within['인구편차'] = (
    within['월평균_일최대인구']
    - within.groupby('District')['월평균_일최대인구'].transform('mean')
)

within_r, within_p = stats.pearsonr(within['인구편차'], within['이송편차'])

print(f'전체 데이터 Pearson r = {raw_r:.3f}, p = {raw_p:.4g}')
print(f'자치구 평균 제거 후 Pearson r = {within_r:.3f}, p = {within_p:.4g}')
```

## 8. 해석용 자동 요약

아래 결과는 **어느 구가 어느 계절에 상대적으로 집중되는지**, 그리고 그 계절에 생활인구가 같이 증가하는지를 빠르게 확인하기 위한 출력입니다.


```python
# 11. 핵심 결과 자동 출력
print('===== 계절성 이송 집중 분석 요약 =====')

for i, (district, row) in enumerate(result.head(10).iterrows(), 1):
    target = row['최대집중계절']
    rep = repeat_df[repeat_df['District'] == district].iloc[0]
    f = focus[focus['District'] == district].iloc[0]

    print(
        f"{i}. {district}: {target} 집중 "
        f"(연평균 비중 {row['최대계절비중']*100:.2f}%, "
        f"25% 대비 +{row['25%초과폭']*100:.2f}%p, "
        f"연도별 반복률 {rep['반복률']*100:.1f}%)"
    )
    print(
        f"   → 해당 계절 이송지수 {f['이송지수']:.1f}, "
        f"생활인구지수 {f['생활인구지수']:.1f}"
    )

print()
print('해석 주의:')
print('- 계절 비중이 높다는 것만으로 원인을 생활인구라고 단정할 수 없습니다.')
print('- 이송지수와 생활인구지수가 함께 상승하는지 확인해야 합니다.')
print('- 이송만 크게 상승하면 기온, 사고, 질환, 관광/유동 특성 등 다른 요인도 검토해야 합니다.')
```

## 9. 분석 결과 저장


```python
# 12. 결과 CSV 저장
display_table.to_csv(
    'district_season_concentration.csv',
    encoding='utf-8-sig'
)
repeat_df.to_csv(
    'district_season_repeatability.csv',
    index=False,
    encoding='utf-8-sig'
)
focus.to_csv(
    'district_season_population_relation.csv',
    index=False,
    encoding='utf-8-sig'
)

print('저장 완료:')
print('- district_season_concentration.csv')
print('- district_season_repeatability.csv')
print('- district_season_population_relation.csv')
```
