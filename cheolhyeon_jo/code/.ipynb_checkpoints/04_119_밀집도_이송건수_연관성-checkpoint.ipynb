## 2. 월별 인구 밀집도 만들기

**원리**
- 밀집도 = 인구 ÷ 면적(km²) : 구 크기 차이를 없앤 비교
- 구 안에서는 면적이 고정 : 밀집도와 인구의 시간 변화 모양이 같음 → 구 간 비교에서 차이가 남
- `전일_` 값의 날짜 = `ds` − 1일 : 실제 그 날의 인구
- 월별 평균 : 20일 이상 자료가 있는 달만 사용 (첫 달처럼 일부만 있는 달 제외)

**밀집도 종류**
- 생활인구밀도 : 하루 평균 인구 기준
- 최대밀집도 : 하루 중 가장 붐빈 시간 인구 기준
- 주간·야간밀집도 : 낮(09-18시)·밤(19-08시) 인구 기준

**면적** : 서울시 행정구역 면적(km²), 최신 통계로 바꿔도 됨

```python
AREA_KM2 = {
    '종로구': 23.91, '중구': 9.96, '용산구': 21.87, '성동구': 16.86, '광진구': 17.06,
    '동대문구': 14.22, '중랑구': 18.50, '성북구': 24.58, '강북구': 23.60, '도봉구': 20.65,
    '노원구': 35.44, '은평구': 29.71, '서대문구': 17.63, '마포구': 23.85, '양천구': 17.41,
    '강서구': 41.44, '구로구': 20.12, '금천구': 13.02, '영등포구': 24.55, '동작구': 16.35,
    '관악구': 29.57, '서초구': 46.98, '강남구': 39.50, '송파구': 33.87, '강동구': 24.59,
}
print('서울 전체 면적 합계 :', round(sum(AREA_KM2.values()), 2), 'km²')

POP_COLS = {
    '전일_total_pop': '생활인구',
    '전일_일최대인구수': '일최대인구',
    '전일_day_pop': '주간인구',
    '전일_night_pop': '야간인구',
}

daily = pop[['ds', 'District'] + list(POP_COLS)].rename(columns=POP_COLS)
daily['date'] = daily['ds'] - pd.Timedelta(days=1)
daily['ym'] = daily['date'].dt.to_period('M').dt.to_timestamp()

monthly = daily.groupby(['District', 'ym']).agg(
    일수=('date', 'nunique'),
    생활인구=('생활인구', 'mean'),
    일최대인구=('일최대인구', 'mean'),
    주간인구=('주간인구', 'mean'),
    야간인구=('야간인구', 'mean'),
).reset_index()
monthly = monthly[monthly['일수'] >= 20].drop(columns='일수')

area = monthly['District'].map(AREA_KM2)
monthly['생활인구밀도'] = monthly['생활인구'] / area
monthly['최대밀집도'] = monthly['일최대인구'] / area
monthly['주간밀집도'] = monthly['주간인구'] / area
monthly['야간밀집도'] = monthly['야간인구'] / area

print(monthly.shape, monthly['ym'].min().date(), '~', monthly['ym'].max().date())
monthly.head()
```

### 2-1. 같은 달 이송 건수와 합치기
- 연관성 분석용 : 같은 달의 밀집도와 이송 건수
- 이송률 = 생활인구 1만 명당 이송 건수 : 인구 규모 효과를 뺀 값

```python
DENS_COLS = ['생활인구밀도', '최대밀집도', '주간밀집도', '야간밀집도']

merged = transport[['ds', 'District', '연도', '월', target]].merge(
    monthly, left_on=['District', 'ds'], right_on=['District', 'ym'], how='inner').drop(columns='ym')
merged['이송률'] = merged[target] / merged['생활인구'] * 10000
merged = merged.sort_values(['District', 'ds']).reset_index(drop=True)

print(merged.shape, merged['ds'].min().date(), '~', merged['ds'].max().date())
merged[['District', 'ds', target, '생활인구', '최대밀집도', '이송률']].head()
```

## 3. 연관성 분석

**상관계수 두 가지**
- 피어슨 r : 직선 관계의 세기 (−1 ~ 1)
- 스피어만 ρ : 순위 기준 관계, 이상값에 덜 민감
- p-value : 우연히 이 정도 상관이 나올 확률, 0.05 미만이면 '우연 아님'

**함수**
- `corr_test` : 두 열의 피어슨·스피어만 계수와 p값, 판정을 한 줄로 정리
- `add_change` : 구별 전월 대비·전년 동월 대비 변화율 (달이 이어질 때만 계산)

```python
def judge(r, p):
    if p < P_LIMIT and abs(r) >= R_LIMIT:
        return '연관 있음'
    if p < P_LIMIT:
        return '약한 연관'
    return '연관 없음'


def corr_test(df, x, y, label=''):
    d = df[[x, y]].dropna()
    r, p = stats.pearsonr(d[x], d[y])
    rho, p_s = stats.spearmanr(d[x], d[y])
    return {'구분': label, 'X': x, 'Y': y, 'n': len(d),
            '피어슨_r': r, '피어슨_p': p, '스피어만_rho': rho, '스피어만_p': p_s, '판정': judge(r, p)}


def add_change(df, cols, months, suffix):
    # months 개월 전 값과 비교한 변화율(%)
    out = df.copy()
    prev = df[['District', 'ds'] + cols].copy()
    prev['ds'] = prev['ds'] + pd.DateOffset(months=months)
    prev = prev.rename(columns={c: c + '_prev' for c in cols})
    out = out.merge(prev, on=['District', 'ds'], how='left')
    for c in cols:
        out[c + suffix] = (out[c] / out[c + '_prev'] - 1) * 100
    return out.drop(columns=[c + '_prev' for c in cols])
```

### 3-1. 구 간 비교 (25개 구)
- 기간 통일 : 25개 구가 모두 있는 달만 사용 (구마다 기간이 다르면 코로나 시기 등이 섞임)
- 구별 평균 : 밀집도 · 이송 건수 · 이송률
- 확인 포인트 : 밀집한 구가 이송 건수가 많은지 / 1인당 이송도 많은지

```python
cnt = merged.groupby('ds')['District'].nunique()
full_months = cnt[cnt == merged['District'].nunique()].index
common = merged[merged['ds'].isin(full_months)]
print(f'공통 기간 : {len(full_months)}개월 ({full_months.min().date()} ~ {full_months.max().date()})')

gu = common.groupby('District')[['생활인구', '생활인구밀도', '최대밀집도', '야간밀집도', target, '이송률']].mean()

between = pd.DataFrame([
    corr_test(gu, '생활인구', target, '구 간'),
    corr_test(gu, '생활인구밀도', target, '구 간'),
    corr_test(gu, '최대밀집도', target, '구 간'),
    corr_test(gu, '야간밀집도', target, '구 간'),
    corr_test(gu, '최대밀집도', '이송률', '구 간'),
    corr_test(gu, '야간밀집도', '이송률', '구 간'),
])
print(between.round(3).to_string(index=False))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
pairs = [('생활인구', target), ('최대밀집도', target), ('최대밀집도', '이송률')]
for ax, (x, y) in zip(axes, pairs):
    sns.regplot(data=gu, x=x, y=y, ax=ax)
    for name, row in gu.iterrows():
        ax.text(row[x], row[y], name, fontsize=8)
    r = stats.pearsonr(gu[x], gu[y])[0]
    ax.set_title(f'{x} vs {y} (r = {r:.2f})')
plt.tight_layout()
plt.savefig(RESULT_DIR / '04_구간_산점도.png')
plt.show()
```

### 3-2. 전체 월 데이터 상관 (히트맵)

### 3-3. 구 안에서의 시간 변화
- 질문 : 같은 구에서 밀집도가 오른 달에 이송도 늘었나?
- 수준 상관 : 구별 월별 값끼리 상관 (추세·계절이 섞임)
- 전월 대비 변화율 상관 : 규모 차이 제거, 계절 영향은 남음
- 전년 동월 대비 변화율 상관 : 규모 차이·계절 영향 모두 제거 → 가장 공정한 비교

```python
chg = add_change(merged, ['최대밀집도', '생활인구밀도', target], 1, '_전월비')
chg = add_change(chg, ['최대밀집도', '생활인구밀도', target], 12, '_전년비')

rows = []
for name, g in chg.groupby('District'):
    for x, y, kind in [('최대밀집도', target, '수준'),
                       ('최대밀집도_전월비', target + '_전월비', '전월 대비 변화'),
                       ('최대밀집도_전년비', target + '_전년비', '전년 동월 대비 변화')]:
        d = g[[x, y]].dropna()
        if len(d) >= 12:
            r, p = stats.pearsonr(d[x], d[y])
            rows.append({'지역구명': name, '비교': kind, 'n': len(d), '피어슨_r': r, '피어슨_p': p})
within = pd.DataFrame(rows)

summary = within.groupby('비교').agg(
    구_수=('지역구명', 'count'),
    r_중앙값=('피어슨_r', 'median'),
    양의_유의한_구=('피어슨_p', lambda s: int(((s < P_LIMIT) & (within.loc[s.index, '피어슨_r'] > 0)).sum())),
    음의_유의한_구=('피어슨_p', lambda s: int(((s < P_LIMIT) & (within.loc[s.index, '피어슨_r'] < 0)).sum())),
)
print(summary.round(3))

pooled = pd.DataFrame([
    corr_test(chg, '최대밀집도_전월비', target + '_전월비', '구 내 (전체 합침)'),
    corr_test(chg, '생활인구밀도_전월비', target + '_전월비', '구 내 (전체 합침)'),
    corr_test(chg, '최대밀집도_전년비', target + '_전년비', '구 내 (전체 합침)'),
    corr_test(chg, '생활인구밀도_전년비', target + '_전년비', '구 내 (전체 합침)'),
])
print()
print(pooled.round(3).to_string(index=False))
```

```python
plot_df = within.pivot(index='지역구명', columns='비교', values='피어슨_r')
plot_df = plot_df[['수준', '전월 대비 변화', '전년 동월 대비 변화']].sort_values('전년 동월 대비 변화')

plot_df.plot(kind='barh', figsize=(10, 10), width=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.axvline(R_LIMIT, color='gray', linestyle='--')
plt.axvline(-R_LIMIT, color='gray', linestyle='--')
plt.xlabel('피어슨 r (최대밀집도 vs 이송 건수)')
plt.title('구별 : 밀집도와 이송 건수의 상관')
plt.tight_layout()
plt.savefig(RESULT_DIR / '04_구내_상관.png')
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, s in zip(axes, ['_전월비', '_전년비']):
    sns.regplot(data=chg, x='최대밀집도' + s, y=target + s, ax=ax,
                scatter_kws={'alpha': 0.3, 's': 10}, line_kws={'color': 'red'})
    ax.set_title(f'최대밀집도{s} vs 이송 건수{s} (%)')
plt.tight_layout()
plt.savefig(RESULT_DIR / '04_변화율_산점도.png')
plt.show()
```

#### 3-3-1. 코로나 시기 영향 확인
- 2020~2021 : 인구 이동과 이송이 함께 크게 줄어든 기간 → 상관을 부풀릴 수 있음
- 기간을 나눠 전년비 상관이 유지되는지 확인

```python
periods = {'2018~2019 (코로나 전)': (2018, 2019), '2020~2021 (코로나)': (2020, 2021), '2022~ (코로나 후)': (2022, 2100)}
period_rows = []
for label, (lo, hi) in periods.items():
    part = chg[(chg['ds'].dt.year >= lo) & (chg['ds'].dt.year <= hi)]
    period_rows.append(corr_test(part, '최대밀집도_전년비', target + '_전년비', label))
period_df = pd.DataFrame(period_rows)
print(period_df[['구분', 'n', '피어슨_r', '피어슨_p', '판정']].round(4).to_string(index=False))
```

### 3-4. 시차 상관 (먼저 움직이는가?)
- 예측에 쓸 수 있는 값 : 이번 달 이전에 알 수 있는 값 → 전월 밀집도
- k개월 전 밀집도 변화와 이번 달 이송 변화의 상관 (k = 0 ~ 3)
- k = 0 : 같은 달 (연관은 보여도 예측 입력으로는 못 씀)
- k ≥ 1 : 상관이 남아 있으면 예측에 쓸 수 있는 신호

```python
lag_rows = []
for s, m in [('_전년비', 12), ('_전월비', 1)]:
    for k in range(0, 4):
        lagged = chg[['District', 'ds', '최대밀집도' + s]].copy()
        lagged['ds'] = lagged['ds'] + pd.DateOffset(months=k)
        d = chg[['District', 'ds', target + s]].merge(lagged, on=['District', 'ds'])
        res = corr_test(d, '최대밀집도' + s, target + s, f'{k}개월 전')
        res.update({'변화율': s.strip('_'), '시차(개월)': k})
        lag_rows.append(res)
lag_df = pd.DataFrame(lag_rows)[['변화율', '시차(개월)', 'n', '피어슨_r', '피어슨_p', '판정']]
print(lag_df.round(3).to_string(index=False))

plt.figure(figsize=(8, 4))
sns.barplot(data=lag_df, x='시차(개월)', y='피어슨_r', hue='변화율')
plt.axhline(0, color='black', linewidth=0.8)
plt.title('k개월 전 밀집도 변화 vs 이번 달 이송 변화')
plt.tight_layout()
plt.savefig(RESULT_DIR / '04_시차_상관.png')
plt.show()
```

### 3-5. 연관성 판정
- 구 간 : 밀집한 구가 이송이 많은가 (공간 연관)
- 구 내 : 밀집도가 변할 때 이송도 같이 변하는가 (시간 연관)
- 예측 신호 : 1개월 전 밀집도 변화가 이번 달 이송 변화와 상관이 있는가

```python
def pick(df, **cond):
    m = pd.Series(True, index=df.index)
    for k, v in cond.items():
        m &= df[k] == v
    return df[m].iloc[0]

space = pick(between, X='최대밀집도', Y=target)
space_rate = pick(between, X='최대밀집도', Y='이송률')
time_yoy = pick(pooled, X='최대밀집도_전년비')
lag1 = pick(lag_df, 변화율='전년비', **{'시차(개월)': 1})

assoc = pd.DataFrame([
    {'관점': '구 간 : 밀집도 vs 이송 건수', 'r': space['피어슨_r'], 'p': space['피어슨_p'], '판정': space['판정']},
    {'관점': '구 간 : 밀집도 vs 이송률', 'r': space_rate['피어슨_r'], 'p': space_rate['피어슨_p'], '판정': space_rate['판정']},
    {'관점': '구 내 : 전년비 변화 (같은 달)', 'r': time_yoy['피어슨_r'], 'p': time_yoy['피어슨_p'], '판정': time_yoy['판정']},
    {'관점': '예측 신호 : 1개월 전 전년비 변화', 'r': lag1['피어슨_r'], 'p': lag1['피어슨_p'], '판정': lag1['판정']},
])
print(assoc.round(4).to_string(index=False))

has_assoc = (assoc['판정'] != '연관 없음').any()
print('\n연관성 :', '있음 → 4장에서 예측 기여 확인' if has_assoc else '없음 (그래도 4장에서 예측 기여를 확인)')

pd.concat([between, pooled, period_df], ignore_index=True).to_csv(RESULT_DIR / '04_상관분석.csv', index=False, encoding='utf-8-sig')
within.to_csv(RESULT_DIR / '04_구내_상관.csv', index=False, encoding='utf-8-sig')
lag_df.to_csv(RESULT_DIR / '04_시차_상관.csv', index=False, encoding='utf-8-sig')
```

## 4. 예측에 도움이 되는가?

**실험 방법 (빼보기 실험)**
- 같은 행, 같은 모델, 같은 평가 기간에서 특성만 바꿔 비교
- 차이가 나면 그 차이 = 추가한 특성의 효과

**특성 세트 3종**
- 기본 : 인구 정보 없이 과거 이송 건수·구조 실적만
- 기본+인구 : 전처리에서 선택한 전월 인구 컬럼 4개 추가 (현재 방식)
- 기본+밀집도 : 전월 밀집도 3개 추가 (이번 분석)

**밀집도 특성 (모두 지난달까지의 값 → 미래 예측에 사용 가능)**
- 전월_생활인구밀도 · 전월_최대밀집도 : 지난달 밀집 수준
- 전월_최대밀집도_전년비 : 지난달 밀집도가 1년 전보다 몇 % 변했는지 (3-4의 예측 신호)

```python
dens_prev = monthly[['District', 'ym', '생활인구밀도', '최대밀집도']].copy()
dens_prev = dens_prev.rename(columns={'ym': 'ds'})
dens_prev = add_change(dens_prev, ['최대밀집도'], 12, '_전년비')
dens_prev['ds'] = dens_prev['ds'] + pd.DateOffset(months=1)      # 지난달 값 → 이번 달 행에 붙임
dens_prev = dens_prev.rename(columns={'생활인구밀도': '전월_생활인구밀도',
                                      '최대밀집도': '전월_최대밀집도',
                                      '최대밀집도_전년비': '전월_최대밀집도_전년비'})

model_df = transport.merge(dens_prev, on=['District', 'ds'], how='left')

POP_FEATS = ['전월_night_pop', '전월_동일자치구행정동간이동인구수', '전월_일최대인구수', '전월_서울외유입인구수']
selected = feat_info.loc[feat_info['결과'] == '선택', '컬럼'].tolist()
BASE_FEATS = ['월'] + [c for c in selected if c not in POP_FEATS]
DENS_FEATS = ['전월_생활인구밀도', '전월_최대밀집도', '전월_최대밀집도_전년비']

FEATURE_SETS = {
    '기본': BASE_FEATS,
    '기본+인구': BASE_FEATS + POP_FEATS,
    '기본+밀집도': BASE_FEATS + DENS_FEATS,
}
for k, v in FEATURE_SETS.items():
    print(f'{k:8s} ({len(v)}개) : {v}')

# 공정 비교 : 모든 세트에 필요한 값이 다 있는 행만
all_feats = sorted(set(sum(FEATURE_SETS.values(), [])))
data = model_df.dropna(subset=all_feats + [target]).sort_values(['ds', 'District']).reset_index(drop=True)
print(f'\n전체 {len(model_df)}행 -> 비교용 {len(data)}행 ({data["ds"].min().date()} ~ {data["ds"].max().date()})')

data[['ds', 'District'] + all_feats + [target]].to_csv(
    ML_DATA_DIR / 'transport_density_dataset.csv', index=False, encoding='utf-8-sig')
```

### 4-1. 모델 정의
- 모델 6종 : LinearRegression · Ridge · DecisionTree · RandomForest · GradientBoosting · DNN
- 표준화(StandardScaler) : 단위가 다른 특성(건수, 명/km², %)을 같은 크기로
- DNN : 정답도 표준화 → 학습 안정, 예측 후 원래 단위로 되돌림
- EarlyStopping : 검증 손실이 더 안 줄면 학습 멈춤 (과적합 방지)
- 정확도 = (1 − MAPE) × 100

```python
MODEL_NAMES = ['LinearRegression', 'Ridge', 'DecisionTree', 'RandomForest', 'GradientBoosting']
if HAS_TF:
    MODEL_NAMES.append('DNN')


def build_model(name, n_features):
    if name == 'LinearRegression':
        return LinearRegression()
    if name == 'Ridge':
        return Ridge(alpha=1.0)
    if name == 'DecisionTree':
        return DecisionTreeRegressor(max_depth=6, random_state=SEED)
    if name == 'RandomForest':
        return RandomForestRegressor(n_estimators=300, max_depth=10, random_state=SEED)
    if name == 'GradientBoosting':
        return GradientBoostingRegressor(n_estimators=300, max_depth=3, learning_rate=0.05, random_state=SEED)
    # DNN : 은닉층 2개 + Dropout
    model = Sequential([
        Input(shape=(n_features,)),
        Dense(64, activation='relu'),
        Dropout(0.2),
        Dense(32, activation='relu'),
        Dense(1),
    ])
    model.compile(optimizer='adam', loss='mse')
    return model


def fit_model(name, X, y):
    x_scaler = StandardScaler().fit(X)
    Xs = x_scaler.transform(X)
    model = build_model(name, X.shape[1])
    y_scaler = None
    if name == 'DNN':
        y_scaler = StandardScaler().fit(y.values.reshape(-1, 1))
        es = EarlyStopping(monitor='val_loss', patience=30, restore_best_weights=True)
        model.fit(Xs, y_scaler.transform(y.values.reshape(-1, 1)),
                  validation_split=0.2, epochs=500, batch_size=32, callbacks=[es], verbose=0)
    else:
        model.fit(Xs, y)
    return {'name': name, 'model': model, 'x_scaler': x_scaler, 'y_scaler': y_scaler}


def predict(bundle, X):
    Xs = bundle['x_scaler'].transform(X)
    if bundle['name'] == 'DNN':
        pred = bundle['model'].predict(Xs, verbose=0)
        return bundle['y_scaler'].inverse_transform(pred).flatten()
    return bundle['model'].predict(Xs)


def evaluate(y_true, y_pred):
    return {
        '정확도(%)': (1 - mean_absolute_percentage_error(y_true, y_pred)) * 100,
        'R2': r2_score(y_true, y_pred),
        'MAE': mean_absolute_error(y_true, y_pred),
        'RMSE': np.sqrt(mean_squared_error(y_true, y_pred)),
    }
```

### 4-2. 특성 세트 3종 × 모델 6종 (최근 1년으로 평가)
- 시간순 분할 : 마지막 12개월 = 평가, 그 이전 = 학습
- MAE 개선율 = (기본 MAE − 세트 MAE) ÷ 기본 MAE × 100 : 양수면 오차가 줄어든 것
- 기준선 : 이번 달 = 지난달 그대로

```python
split_date = data['ds'].max() - pd.DateOffset(months=11)
train = data[data['ds'] < split_date]
test = data[data['ds'] >= split_date]
y_train, y_test = train[target], test[target]
print(f'학습 {len(train)}행 / 평가 {len(test)}행 (평가 시작 {split_date.date()})')

scores, preds, bundles = [], {}, {}
for set_name, feats in FEATURE_SETS.items():
    for name in MODEL_NAMES:
        b = fit_model(name, train[feats], y_train)
        p = predict(b, test[feats])
        bundles[(set_name, name)] = b
        preds[(set_name, name)] = p
        scores.append({'특성세트': set_name, '모델': name, **evaluate(y_test, p)})
    print(f'{set_name} 완료')

score_df = pd.DataFrame(scores)
base_mae = score_df[score_df['특성세트'] == '기본'].set_index('모델')['MAE']
score_df['MAE_개선율(%)'] = (score_df['모델'].map(base_mae) - score_df['MAE']) / score_df['모델'].map(base_mae) * 100

naive = evaluate(y_test, test['전월_transport_cnt'])
print(f"\n기준선(전월 그대로) : 정확도 {naive['정확도(%)']:.2f}% / MAE {naive['MAE']:.1f}")
print(score_df.round(3).to_string(index=False))
score_df.to_csv(RESULT_DIR / '04_모델비교.csv', index=False, encoding='utf-8-sig')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))
sns.barplot(data=score_df, x='모델', y='MAE', hue='특성세트', ax=axes[0])
axes[0].axhline(naive['MAE'], color='gray', linestyle='--', label='기준선(전월 그대로)')
axes[0].set_title('모델 × 특성 세트 MAE (낮을수록 좋음)')
axes[0].legend()
axes[0].tick_params(axis='x', rotation=30)

sns.barplot(data=score_df[score_df['특성세트'] != '기본'], x='모델', y='MAE_개선율(%)', hue='특성세트', ax=axes[1])
axes[1].axhline(0, color='black', linewidth=0.8)
axes[1].set_title('기본 대비 MAE 개선율 (양수 = 도움)')
axes[1].tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.savefig(RESULT_DIR / '04_모델비교.png')
plt.show()
```

### 4-3. 연도별 반복 검증 (우연인지 확인)
- 평가 1년만으로는 운일 수 있음 → 평가 연도를 바꿔가며 반복
- 확장 윈도우 : 평가 연도 이전 전체로 학습, 그 해 12개월로 평가
- 학습이 최소 2년 이상인 해만 평가
- 시간 절약 : DNN 제외, 머신러닝 5종만
- 판단 기준 : 개선된 경우 비율(승률) · 평균 개선율

```python
CV_MODELS = [m for m in MODEL_NAMES if m != 'DNN']
years = sorted(data['ds'].dt.year.unique())
first_year = data['ds'].dt.year.min()
test_years = [y for y in years if y - first_year >= 2 and (data['ds'].dt.year == y).sum() >= 100]
print('평가 연도 :', test_years)

cv_rows = []
for year in test_years:
    tr = data[data['ds'].dt.year < year]
    te = data[data['ds'].dt.year == year]
    for set_name, feats in FEATURE_SETS.items():
        for name in CV_MODELS:
            p = predict(fit_model(name, tr[feats], tr[target]), te[feats])
            cv_rows.append({'평가연도': year, '특성세트': set_name, '모델': name,
                            'MAE': mean_absolute_error(te[target], p)})
cv = pd.DataFrame(cv_rows)

cv_wide = cv.pivot_table(index=['평가연도', '모델'], columns='특성세트', values='MAE').reset_index()
for s in ['기본+인구', '기본+밀집도']:
    cv_wide[s + '_개선율(%)'] = (cv_wide['기본'] - cv_wide[s]) / cv_wide['기본'] * 100

cv_summary = pd.DataFrame([{
    '특성세트': s,
    '비교_횟수': len(cv_wide),
    '개선_횟수': int((cv_wide[s + '_개선율(%)'] > 0).sum()),
    '승률(%)': (cv_wide[s + '_개선율(%)'] > 0).mean() * 100,
    '평균_개선율(%)': cv_wide[s + '_개선율(%)'].mean(),
    '중앙_개선율(%)': cv_wide[s + '_개선율(%)'].median(),
} for s in ['기본+인구', '기본+밀집도']])
print(cv_summary.round(2).to_string(index=False))
cv_wide.to_csv(RESULT_DIR / '04_연도별검증.csv', index=False, encoding='utf-8-sig')
```

```python
plot_cv = cv_wide.melt(id_vars=['평가연도', '모델'], value_vars=['기본+인구_개선율(%)', '기본+밀집도_개선율(%)'],
                       var_name='특성세트', value_name='개선율(%)')
plot_cv['특성세트'] = plot_cv['특성세트'].str.replace('_개선율(%)', '', regex=False)

g = sns.catplot(data=plot_cv, x='평가연도', y='개선율(%)', hue='특성세트', col='모델',
                kind='bar', col_wrap=3, height=3.5, aspect=1.2)
for ax in g.axes.flat:
    ax.axhline(0, color='black', linewidth=0.8)
g.fig.suptitle('연도별 MAE 개선율 (기본 대비, 양수 = 도움)', y=1.02)
plt.savefig(RESULT_DIR / '04_연도별검증.png', bbox_inches='tight')
plt.show()
```

### 4-4. 순열 중요도 (모델이 밀집도를 실제로 쓰는가?)
- 원리 : 한 특성의 값만 무작위로 섞기 → 오차가 얼마나 커지는지 측정
- 오차 증가가 큼 = 모델이 그 특성에 많이 의존
- 0 근처 = 없어도 되는 특성
- 대상 : 4-2의 `기본+밀집도` 세트에서 MAE가 가장 낮은 모델

```python
def permutation_mae(bundle, X, y, n_repeats=20):
    rng = np.random.default_rng(SEED)
    base = mean_absolute_error(y, predict(bundle, X))
    rows = []
    for col in X.columns:
        ups = []
        for _ in range(n_repeats):
            Xp = X.copy()
            Xp[col] = rng.permutation(Xp[col].values)
            ups.append(mean_absolute_error(y, predict(bundle, Xp)) - base)
        rows.append({'특성': col, 'MAE_증가_평균': np.mean(ups), 'MAE_증가_표준편차': np.std(ups)})
    return pd.DataFrame(rows).sort_values('MAE_증가_평균', ascending=False).reset_index(drop=True)


dens_scores = score_df[score_df['특성세트'] == '기본+밀집도'].sort_values('MAE')
best_name = dens_scores.iloc[0]['모델']
print('대상 모델 :', best_name)

imp = permutation_mae(bundles[('기본+밀집도', best_name)], test[FEATURE_SETS['기본+밀집도']], y_test)
imp['밀집도_특성'] = imp['특성'].isin(DENS_FEATS)
print(imp.round(3).to_string(index=False))
imp.to_csv(RESULT_DIR / '04_순열중요도.csv', index=False, encoding='utf-8-sig')

plt.figure(figsize=(9, 5))
colors = ['tomato' if d else 'steelblue' for d in imp['밀집도_특성']]
plt.barh(imp['특성'], imp['MAE_증가_평균'], xerr=imp['MAE_증가_표준편차'], color=colors)
plt.gca().invert_yaxis()
plt.xlabel('섞었을 때 MAE 증가량 (건)')
plt.title(f'{best_name} 순열 중요도 (빨강 = 밀집도 특성)')
plt.tight_layout()
plt.savefig(RESULT_DIR / '04_순열중요도.png')
plt.show()
```

### 4-5. 구별 효과
- 같은 모델에서 `기본` vs `기본+밀집도`의 구별 MAE 비교
- 전체 평균이 비슷해도 특정 구에서만 도움이 될 수 있음

```python
by_gu = test[['District', target]].copy()
by_gu['기본_오차'] = (by_gu[target] - preds[('기본', best_name)]).abs()
by_gu['밀집도_오차'] = (by_gu[target] - preds[('기본+밀집도', best_name)]).abs()
by_gu = by_gu.groupby('District')[['기본_오차', '밀집도_오차']].mean()
by_gu['개선율(%)'] = (by_gu['기본_오차'] - by_gu['밀집도_오차']) / by_gu['기본_오차'] * 100
by_gu = by_gu.sort_values('개선율(%)').reset_index().rename(columns={'District': '지역구명'})

print(f"개선된 구 : {(by_gu['개선율(%)'] > 0).sum()} / {len(by_gu)}")
by_gu.to_csv(RESULT_DIR / '04_구별_개선.csv', index=False, encoding='utf-8-sig')

plt.figure(figsize=(9, 8))
colors = ['tomato' if v > 0 else 'gray' for v in by_gu['개선율(%)']]
plt.barh(by_gu['지역구명'], by_gu['개선율(%)'], color=colors)
plt.axvline(0, color='black', linewidth=0.8)
plt.xlabel('MAE 개선율 (%)')
plt.title(f'{best_name} : 밀집도 추가 후 구별 개선율 (빨강 = 개선)')
plt.tight_layout()
plt.savefig(RESULT_DIR / '04_구별_개선.png')
plt.show()
```

## 5. 결론과 다음 단계

**판정 규칙**
- 연관성 : p < 0.05 이고 |r| ≥ 0.3 → 연관 있음
- 예측 도움 : 연도별 검증 승률 ≥ 60% 이고 평균 개선율 ≥ 1% → 도움 됨
- 승률 > 50% 이고 평균 개선율 > 0 → 약간 도움 (특성은 넣되 기대는 낮게)
- 그 외 → 도움 안 됨 (특성을 늘리면 복잡도만 증가)

**저장 파일**
- `ml_data/features_density.csv` : 밀집도 특성의 선택/제외와 이유 (예측 노트북에서 사용)
- `ml_data/transport_density_dataset.csv` : 밀집도 특성이 붙은 학습 데이터
- `ml_result/04_*.csv, 04_*.png` : 분석 표와 그래프

```python
def help_level(row):
    if row['승률(%)'] >= 60 and row['평균_개선율(%)'] >= 1:
        return '도움 됨'
    if row['승률(%)'] > 50 and row['평균_개선율(%)'] > 0:
        return '약간 도움'
    return '도움 안 됨'


cv_summary['판정'] = cv_summary.apply(help_level, axis=1)
dens_row = cv_summary[cv_summary['특성세트'] == '기본+밀집도'].iloc[0]
pop_row = cv_summary[cv_summary['특성세트'] == '기본+인구'].iloc[0]
dens_imp = imp[imp['밀집도_특성']]

print('=' * 60)
print('[1] 연관성')
for _, r in assoc.iterrows():
    print(f"  - {r['관점']:30s} r = {r['r']:+.3f} (p = {r['p']:.4f}) → {r['판정']}")
print('\n[2] 예측 기여 (연도별 반복 검증)')
for _, r in cv_summary.iterrows():
    print(f"  - {r['특성세트']:8s} 승률 {r['승률(%)']:.0f}% / 평균 개선율 {r['평균_개선율(%)']:+.2f}% → {r['판정']}")
print(f"\n[3] 최근 1년 최고 모델({best_name}) 밀집도 특성 순열 중요도")
for _, r in dens_imp.iterrows():
    print(f"  - {r['특성']:22s} MAE +{r['MAE_증가_평균']:.2f}건")
print('=' * 60)

use_density = dens_row['판정'] != '도움 안 됨'
reason = f"연도별 검증 승률 {dens_row['승률(%)']:.0f}%, 평균 MAE 개선율 {dens_row['평균_개선율(%)']:+.2f}%"
decision = pd.DataFrame([{
    '컬럼': c,
    '결과': '선택' if use_density else '제외',
    '판정': dens_row['판정'],
    '이유': reason,
} for c in DENS_FEATS])
decision.to_csv(ML_DATA_DIR / 'features_density.csv', index=False, encoding='utf-8-sig')
cv_summary.to_csv(RESULT_DIR / '04_결론.csv', index=False, encoding='utf-8-sig')
print(decision.to_string(index=False))
```

**해석 가이드**
- 구 간 상관은 강하고 구 내 변화 상관은 약한 경우 : 밀집도는 "어느 구가 많은가"를 설명, "다음 달 얼마나 변하나"는 설명 못 함
- 이송 건수 모델은 `전월_transport_cnt`가 이미 구의 규모를 담고 있음 → 밀집도 수준 값은 겹치는 정보
- 밀집도 변화(전년비)가 예측 신호가 되는지는 3-4 시차 상관과 4-4 순열 중요도로 함께 확인
- 코로나 시기(2020~2021) : 인구 이동과 이송이 함께 줄어든 기간, 상관을 키울 수 있음

- 이유 : `전월_transport_cnt`가 구의 규모와 최근 흐름을 이미 담고 있어 밀집도가 새 정보를 거의 못 더함
